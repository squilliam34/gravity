**Test a cluster-weighted portfolio**
Each cluster leader is weighted by the aggregate market capitalization of the cluster it represents. The workflow below mirrors the equal-weighted portfolio analysis so the results are directly comparable.

```python
import sys
sys.path.append('../')
from IPython.display import display
from utils import *
from src.portfolio.portfolio_construction import construct_portfolio
```

**Rebalance every 6 months**

```python
portfolios = []
for year in range(2010, 2026, 5):
    portfolios.append(
        construct_portfolio(
            year=year, 
            freq='6MS', 
            schema='cluster', 
            benchmark='^GSPC'
        )
    )
```

```python
for portfolio in portfolios:
    portfolio.load_prices()
    portfolio.plot()
    start_date = portfolio.start_date.strftime('%Y-%m-%d')
    end_date = portfolio.end_date.strftime('%Y-%m-%d')
    display(
        portfolio.portfolio_performance(
            risk_free_rate=get_risk_free_rate(
                start_date=start_date,
                end_date=end_date,
            )
        )
    )
```
