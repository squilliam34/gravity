**Test an equal weighting portfolio**
- This portfolio is initialized using an equal weighting schema of cluster leaders at each point in time

```python
import sys
sys.path.append('../')
from utils import *
from src.portfolio.portfolio_construction import construct_portfolio
```

```python
portfolios = []
for year in range(2010, 2026, 5):
    portfolios.append(
        construct_portfolio(
            year=year, 
            freq='6MS', 
            schema='equal', 
            benchmark='^GSPC'
        )
    )
```

```python
for portfolio in portfolios:
    portfolio.load_prices()
    portfolio.plot()
    start_date = portfolio.start_date.strftime('%Y-%m-%d')
    end_date = portfolio.end_date.strftime('%Y-%m-%d')
    print(
        f'Sharpe ratio for {start_date} - {end_date}:' 
        f'{portfolio.sharpe(
            risk_free_rate=get_risk_free_rate(
                start_date=start_date,
                end_date=end_date
            )
        )}')
```
