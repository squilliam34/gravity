**Test an equal weighting portfolio**
This portfolio is initialized using an equal weighting schema of cluster leaders at each point in time.

Portfolios consistenly performed best in the first period from 2010-2015, regardless of rebalance period. My initial thought was survivorship bias (tickers only include surviving tickers that didn't get restructured/acquired), however:
1) I highly doubt that companies in poor financial health would be identified as cluster leaders on account of having lower market caps.
2) The portfolio for the most recent period from 2025 on consistently had the second highest Sharpe ratio. Although it did not consistently have the second highest cumulative outperformance or annualized alpha, which would indicate that it experienced lower volatility.
3) When rebalancing every 6 months and every month, the equal-weighted portfolio slightly outperforms the S&P 500 in 2020-2025 by about 6% and 3% respectively. 
4) Finally, it is possible that the early outperformance may be an interest rate regime story. Following the GFC, interest rates were kept lower. The factor model I used as part of the distance component of gravity within a cluster makes use of interest rate exposure and momentum exposure. I wonder if there was interaction between factor exposures and the low interest rate environment of the early 2010s. However, I would also expect to see consistent outperformance in the 2015-2020 period if this were the case, which I do not see either.
5) Even in periods where I do not outperform the benchmark using an equal-weighted portfolio, I still track the performance of the S&P500 and maintain competitive risk-adjusted returns, suggesting the network construction may provide diversification benefits rather than purely return generation.
